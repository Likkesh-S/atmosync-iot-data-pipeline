# AtmoSync Environmental Telemetry Analysis

This notebook loads telemetry data, cleans it, calculates hourly metrics, and visualizes temperature trends.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/telemetry_data.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])
df.head()
```

```python
# Basic data quality checks
print('Rows:', len(df))
print('Missing values:')
print(df.isnull().sum())
print('Duplicates:', df.duplicated().sum())
```

```python
# Hourly metrics
df['telemetry_hour'] = df['timestamp'].dt.floor('h')
hourly = df.groupby('telemetry_hour').agg(
    ping_count=('device_id', 'count'),
    avg_temp_c=('temperature_c', 'mean'),
    peak_temp_c=('temperature_c', 'max'),
    min_temp_c=('temperature_c', 'min'),
    avg_humidity_pct=('humidity_pct', 'mean'),
    avg_pressure_hpa=('pressure_hpa', 'mean')
).reset_index()
hourly.round(2)
```

```python
# Temperature visualization
plt.figure(figsize=(10, 5))
plt.plot(df['timestamp'], df['temperature_c'], marker='o')
plt.title('AtmoSync Temperature Trend')
plt.xlabel('Timestamp')
plt.ylabel('Temperature (°C)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
